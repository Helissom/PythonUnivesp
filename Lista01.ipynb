Objetivo: criar três variáveis com tipos de dados diferentes.

Em Python, uma variável armazena um valor. O tipo de dado depende do valor atribuído.

```python
nome = 'Eliahu'  # str: texto
ano = 1994       # int: número inteiro
valor = 19.01    # float: número decimal

print(nome, ano, valor)
```

Tipos utilizados:

str: textos, escritos entre aspas.

int: números inteiros, sem casas decimais.

float: números com casas decimais

8. Solicitar um número e convertê-lo para float
Objetivo: receber um valor digitado, convertê-lo para número decimal e exibir seu valor e tipo.

```python
'''Uma vez criada nossa variável num, com seu conteúdo vindo da interação com o
usuário, podemos atualizar o conteúdo dessa variável, mudando inclusive seu tipo
de dado.
'''
num = input('Digite um número: ')
num = float(num)

print(num)
print(type(num))
```

input() recebe o valor digitado como texto.

float(num) converte o texto para um número decimal.

print(num) exibe o valor convertido.

print(type(num)) mostra o tipo de dado.
